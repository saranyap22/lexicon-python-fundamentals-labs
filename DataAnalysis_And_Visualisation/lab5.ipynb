BIKE RENTAL CLEANUP

```python
import pandas as pd

bike_rentals_raw = pd.read_csv('data/bike_rentals.csv')
```

```python
# Keep orginal Data safe
rentals = bike_rentals_raw.copy()

# view no of column and now or rows in dataframe
print(rentals.shape)

# what data type each column is
print(rentals.dtypes)
print(rentals.info())

#what data type does price have
rentals['price'].dtype
```

```python
# Find columns where every columns are missing
rentals.isna().sum()
```

```python
# Check if all the coloumn empty for specified columns
print(rentals[['coupon_code','damage_notes','late_fee_note']].isna().sum())

# Check what are the columns are empty
empty = rentals.columns[rentals.isna().all()]
print(len(empty))
print(list(empty))

# Remove all empty columns (coupon_code, damage_notes,late_fee_notes)
# rentals.dropna() wrong syntax every column will be delete
rentals[['coupon_code','damage_notes','late_fee_note']].dropna(axis=1,how='all')

# check shape to see if column removed
print(rentals[['coupon_code','damage_notes','late_fee_note']])
rentals.shape
```
