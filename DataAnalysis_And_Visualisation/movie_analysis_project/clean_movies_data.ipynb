# Clean movies data set

```python
import pandas as pd
movies_raw = pd.read_csv('data/movies.csv')
movies = movies_raw.copy()
```

Identify columns for cleanup
1. Any column has data type mismatch. Convert it. Check before and after count.
2. Any column has all values empty? then remove it.
3. Any column has maximum missing values, then check if any reason for missing? After that decide if you want to remove the coloumn or fill the missing values with madeup data or without removing existing column put values in new column.

```python
print(movies_raw['Release Date'].dtype)
print(movies_raw['Release Date'].isna().sum())
movies['Release Date'] = pd.to_datetime(movies['Release Date'],errors='coerce')
movies['Release Date'].dtype
print(movies['Release Date'].isna().sum())
```

```python
print(movies.loc[6,'Release Date'])
print(movies['Release Date'].count())
movies['Release Date'].sort_values(ascending=False).head(20)
movies.loc[85,'Title']
movies.groupby(movies['Release Date'].dt.year)['Release Date'].count().tail(8)
```

```python
movies['Release Date'].describe()
movies_incorrect_date_mask = movies['Release Date'].dt.year >= 2011
movies_correct_date_mask = movies['Title'] != 'The Adventures of Tintin: Secret of the Unicorn'
combined_date_mask = movies_incorrect_date_mask & movies_correct_date_mask
print('incorrect',(movies_incorrect_date_mask).sum())
print('correct',(movies_correct_date_mask).sum())
print('combined',(combined_date_mask).sum())
movies[(movies['Release Date'].dt.year >= 2011)][['Title','Release Date','Director']].sort_values('Release Date',ascending=False)
```

```python
years_offset = pd.DateOffset(years=100)
movies.loc[combined_date_mask,'Release Date'] = movies.loc[combined_date_mask,'Release Date'] - years_offset
print(movies.groupby(movies['Release Date'].dt.year)['Release Date'].count())
```

```python

print(movies['Release Date'].max())
print(movies[(movies['Release Date'].dt.year >= 2011)][['Title','Release Date']].sort_values('Release Date',ascending=False))
print(movies.groupby(movies['Release Date'].dt.year)[['Release Date']].count().tail(10))
print(movies[movies['Title'] == 'The Adventures of Tintin: Secret of the Unicorn'])
movies[(movies['Release Date'].dt.year >= 2011)][['Title','Release Date','Director']].sort_values('Release Date',ascending=False)
```
