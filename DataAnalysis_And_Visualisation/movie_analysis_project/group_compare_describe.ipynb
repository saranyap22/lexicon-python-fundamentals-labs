# Group Compare and Describe


```python
import pandas as pd
movies = pd.read_csv('data/movies_merged.csv',parse_dates=['Release Date'])
print(movies['Release Date'].dtype,movies['Release Year'].dtype)
```

How many number of films released per each year?
1. Identify column to look up
2. Identigy groupby value

```python
movies.groupby('Release Year')['Title'].count().sort_values(ascending=False)
```

Observation:
1. The most number of movies in the data set is in 2006 and 219 movies

Which Genre group earn most Worlwide typically?
1. Identify the column to groupby
2. Identify the column to calculate mean and median
3. Choose mean or median suits better?

```python
(movies.groupby('Genre group')['Worldwide Gross']).describe().round(0).sort_values('50%',ascending=False)
```

Observation:
1. The difference between mean and median is huge. So median is better for finding typical Worldwide Gross for Genre group.
2. In the data set Action & adventure is the Genre group with higest typical Worldwide earning of 89 million dollar approx


Which Genre group has the most profit and most loss?
1. Group by 'Genre group', 'Profit Status'

```python
movies.groupby(['Genre group','Profit Status'])['Title'].count().unstack().sort_values(['Profit','Loss'],ascending=False)
```

Observation:
1. The most no of Profit making Genre group is Comedy & music.
2. The most no of Loss making Genre group is Drama
3. Check the above observation in terms of share

```python
import numpy as np
genre_group_profit_loss = movies.groupby(['Genre group','Profit Status'])['Title']
movies_profit_loss = movies[~movies['Profit'].isna()]
movies_profit_loss['is_profit'] = np.where(movies_profit_loss['Profit Status'] == 'Profit', True,False)
movies_profit_loss.groupby('Genre group')['is_profit'].mean().round(2).sort_values(ascending=False)
movies_profit_loss.groupby('Genre group')['is_profit'].agg(['count','mean']).round(2)
```

Observation:
1. The highest share of films making profit comes from Genre Action & adventure and Thriller & horror (73%) as per the given data set
2. The highest share of films making Loss comes from Drama ((1-0.57 )* 100= 43%) as per the given data set
3. The share of films making profit  comes from Genre Non-fiction is (58%). But that is coming from very few films (48) compared to others
