# Reading Movies Data


### Load the data from file
1. import pandas for data processing.
2. Remember to choose the kernal.
3. read data from data/movies/movies.csv (read_csv())
4. Always keep backup of raw data before doing anything on the data. (copy())
5. For now none of the column cleaned up.(so no need to parse dates parameters (parse_dates())

```python
import pandas as pd
movies_raw = pd.read_csv('data/movies.csv')
movies = movies_raw.copy()
```

### What the data numbers says?
1. Row and column count (shape)
2. data types of each column (info())
3. How many columns are having all the values empty?
4. how many values are empty in each column?

```python
print(movies.shape)
movies.info()
print(movies.isna().sum().sort_values(ascending=False))
print(movies.isna().all().sum())
movies.isna().mean().sort_values(ascending=False).round(2)
```

Observation:
1. Total rows 3201, columns 16
2. float(8), str(8)
3. Date in str format(Release date)
